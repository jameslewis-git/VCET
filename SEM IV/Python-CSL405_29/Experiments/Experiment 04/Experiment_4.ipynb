```python
from pathlib import Path

# Get all files in the current directory
files = [f.name for f in Path('.').iterdir() if f.is_file()]

# Print the list of files
print("Files in the current directory:")
for file in files:
    print(file)

'''
The above script uses the pathlib module to perform the required function.
It replaces the os module's listdir() and isfile() with Path() and iterdir().
'''
```

```python
import os

files = [f for f in os.listdir('.') if os.path.isfile(f)]

print("Files in the current directory:")
for file in files:
    print(file)

'''
The above scipt uses os module to perform the required function.
Utilize the pathlib module to perform the same functionality.
'''
```
